RAG IMPLEMENTATION WITH OWN TEXT

step 1 preparing document for our text

```python

Documents=[
    Document(page_content="""
    Wikipedia is a free online encyclopedia written and maintained by a community of volunteers, known as Wikipedians, through open collaboration and the wiki software MediaWiki.
    """),

    Document(page_content="""
    Python was created by Guido van Rossum
    in 1991.
    The language name is inspired by the British comedy series Monty
    Python.
    Python 3 was released in 2008 with major improvements.
    """),

    Document(page_content="""
    Popular Python libraries include NumPy
    for numerical
    computing, Pandas for data analysis, Django for web frameworks,
    and TensorFlow
    for machine learning.
    """),
]
```

step 2 splitting the document

```python
spliter=RecursiveCharacterTextSplitter(
    #CHUNK_SIZE=20,
    chunk_overlap=30
)
chunks=spliter.split_documents(Documents)
chunks
```

create embeding for chunks

```python
embeddings=GoogleGenerativeAIEmbeddings()
```

```python
embeddings.embed_query("what is cricket")
```

create vector store

```python
vector_store=chroma.from_documents(chunks,embeddings)
```

step 5 semantic search

```python
context=vector_store.similarity_search("what is python",A=3)
```

talk to llm 

```python

response=llm_gemini.invoke(f"what is python ? you can use the following context{context}")
response.content
```

another way of searching

```python
#creating retrieval
retriever=vector_store.as_retriever(
    search_kwargs={"k":2}
)
#test retriveal
query="what is python"
retrieved=retriever.invoke(query)
```
